```python
df['3P%'] = df['3P%'].fillna(0.0)
df['FT%'] = df['FT%'].fillna(0.0)
df['FG%'] = df['FG%'].fillna(0.0)
```

```python
df['eFG%'] = (df['FG'] + (0.5 * df['3P'])) / df['FGA']
```

```python
df['eFG%'] = df['eFG%'].fillna(0.0)
```
